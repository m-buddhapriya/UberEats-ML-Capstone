```python
# Load model comparsion
demand_models = pd.read_csv(
    "../outputs/day5/model_comparison.csv"
)

display(demand_models)
```

```python
# Identify peak demand periods
hourly_demand = pd.read_csv(
    "../data/processed/hourly_demand.csv"
)

hourly_demand["order_timestamp"] = pd.to_datetime(
    hourly_demand["order_timestamp"]
)

hourly_demand["hour"] = (
    hourly_demand["order_timestamp"].dt.hour
)

hourly_profile = (
    hourly_demand.groupby("hour")["order_count"]
    .mean()
    .sort_values(ascending=False)
)

display(hourly_profile.head(10))
```

```python
# Find peak ordering hours
hourly_profile = (
    hourly_demand
    .groupby("hour")["order_count"]
    .agg(["mean", "sum", "max"])
    .sort_values("mean", ascending=False)
)

display(hourly_profile)
```

```python
peak_hours = hourly_profile.head(5)

print("Top 5 peak demand hours:")
display(peak_hours)
```

```python
# Find the lowest-demand hours
low_demand_hours = (
    hourly_profile
    .sort_values("mean", ascending=True)
    .head(5)
)

print("Lowest-demand hours:")
display(low_demand_hours)
```

```python
# Analyze weekday vs weekend demand
print(hourly_demand.columns.tolist())
```

```python
hourly_demand["order_timestamp"] = pd.to_datetime(
    hourly_demand["order_timestamp"]
)
```

```python
# Create day_of_week
hourly_demand["day_of_week"] = (
    hourly_demand["order_timestamp"].dt.dayofweek
)

hourly_demand["day_name"] = (
    hourly_demand["order_timestamp"].dt.day_name()
)

hourly_demand["hour"] = (
    hourly_demand["order_timestamp"].dt.hour
)

hourly_demand["is_weekend"] = (
    hourly_demand["day_of_week"] >= 5
)
```

```python
weekend_profile = (
    hourly_demand
    .groupby("is_weekend")["order_count"]
    .agg(["mean", "sum", "count"])
    .reset_index()
)

weekend_profile["day_type"] = weekend_profile["is_weekend"].map({
    False: "Weekday",
    True: "Weekend"
})

weekend_profile = weekend_profile[
    ["day_type", "mean", "sum", "count"]
]

display(weekend_profile)
```

```python
# Analyze demand by day
daily_profile = (
    hourly_demand
    .groupby(["day_of_week", "day_name"])["order_count"]
    .agg(["mean", "sum"])
    .reset_index()
    .sort_values("day_of_week")
)

display(daily_profile)
```

```python
# Analyse demand by hour
hourly_profile = (
    hourly_demand
    .groupby("hour")["order_count"]
    .agg(["mean", "sum"])
    .reset_index()
)

display(hourly_profile)
```

```python
peak_hours = (
    hourly_profile
    .sort_values("mean", ascending=False)
    .head(5)
)

display(peak_hours)
```

```python
# Identify the busiest day
peak_day = daily_profile.loc[
    daily_profile["mean"].idxmax()
]

print("Highest average-demand day:")
print(peak_day)
print("\n")
# Identify the busiest hour
peak_hour = hourly_profile.loc[
    hourly_profile["mean"].idxmax()
]
print("Highest average-demand hour:")
print(peak_hour)
```

```python
# Create a business-ready demand insight
weekday_mean = weekend_profile.loc[
    weekend_profile["day_type"] == "Weekday",
    "mean"
].iloc[0]

weekend_mean = weekend_profile.loc[
    weekend_profile["day_type"] == "Weekend",
    "mean"
].iloc[0]

if weekend_mean > weekday_mean:
    weekend_result = "Weekend demand is higher than weekday demand."
elif weekend_mean < weekday_mean:
    weekend_result = "Weekday demand is higher than weekend demand."
else:
    weekend_result = "Weekend and weekday demand are approximately equal."

print(weekend_result)
```

```python
# Business Insight Table
demand_insights = pd.DataFrame({
    "Metric": [
        "Weekday average hourly demand",
        "Weekend average hourly demand",
        "Highest-demand day",
        "Highest-demand day average",
        "Peak hour",
        "Peak hour average demand"
    ],
    "Value": [
        round(weekday_mean, 3),
        round(weekend_mean, 3),
        daily_profile.loc[daily_profile["mean"].idxmax(), "day_name"],
        round(daily_profile["mean"].max(), 3),
        f"{int(peak_hour['hour']):02d}:00",
        round(peak_hour["mean"], 3)
    ]
})

display(demand_insights)
```

# Business Recommendations

### 1. Improve Delivery-Time Performance

Delivery delays are the largest identified complaint theme, while delivery
distance is the strongest delivery-time predictor.

The business should prioritize rider allocation for longer-distance orders
and monitor preparation delays at restaurants. This can help reduce the
likelihood of late deliveries and improve customer satisfaction.

### 2. Improve Food Freshness and Temperature

Food quality/freshness and food arriving cold represent major sources of
customer dissatisfaction.

Restaurants and delivery operations should monitor preparation-to-pickup
time and pickup-to-delivery time, particularly for orders involving longer
delivery distances.

### 3. Target Capacity to High-Demand Periods

Demand analysis shows that Thursday has the highest average daily demand,
while 09:00 and 17:00 are among the strongest hourly demand periods.

Additional delivery capacity should therefore be considered during these
historically high-demand periods.

### 4. Use Customer Segments for Targeted Strategies

Customer segmentation can be used to distinguish high-value and lower-
engagement customers.

High-value segments can be targeted with retention and loyalty initiatives,
while lower-engagement segments can be targeted with campaigns designed to
increase ordering frequency and retention.

### 5. Prioritize Underperforming Restaurant Segments

Restaurant clustering provides groups with different order volumes, revenue,
ratings, preparation times, and cancellation rates.

Restaurant groups showing weaker operational performance should receive
greater attention through preparation-time monitoring, service-quality
improvement, and operational support.

# Model Limitations

### Customer and Restaurant Segmentation

The clustering results depend on the selected features and clustering
algorithm. Different feature scaling, cluster counts, or algorithms could
produce different segment definitions.

### Sentiment Analysis

The TF-IDF + Naive Bayes model performed poorly on the available test set,
while VADER performed substantially better. The review dataset also contains
many repeated review templates, which may limit the generalization of the
results to naturally written customer feedback.

### Delivery-Time Prediction

The delivery model achieves strong predictive performance, but the model
is based on historical data. Important real-world factors that are not
captured in the dataset may affect delivery time.

### Demand Forecasting

The forecasting dataset contains relatively low hourly order volumes.
Although Prophet slightly outperformed ARIMA on the evaluation period,
forecast accuracy can change when demand patterns shift.

### General Limitation

The analysis is based on the available synthetic/historical dataset.
Therefore, the findings should be validated against production data before
being used for operational decisions.
