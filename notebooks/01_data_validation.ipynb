# DATA VALIDATION

```python
# Import Libraries
import pandas as pd
import numpy as np
import os
```

```python
# Load Datasets
customers = pd.read_csv("../data/raw/customers.csv")
restaurants = pd.read_csv("../data/raw/restaurants.csv")
drivers = pd.read_csv("../data/raw/drivers.csv")
orders = pd.read_csv("../data/raw/orders.csv")
deliveries = pd.read_csv("../data/raw/deliveries.csv")
reviews = pd.read_csv("../data/raw/reviews.csv")
payments = pd.read_csv("../data/raw/payments.csv")
```

```python
# Check the number of rows
print("Customers:", customers.shape)
print("Restaurants:", restaurants.shape)
print("Drivers:", drivers.shape)
print("Orders:", orders.shape)
print("Deliveries:", deliveries.shape)
print("Reviews:", reviews.shape)
print("Payments:", payments.shape)
```

```python
# Check for columns
print("Customers:", customers.shape)
print("Restaurants:", restaurants.shape)
print("Drivers:", drivers.shape)
print("Orders:", orders.shape)
print("Deliveries:", deliveries.shape)
print("Reviews:", reviews.shape)
print("Payments:", payments.shape)
```

```python
# Check for missing values
print("Customers missing values:")
print(customers.isnull().sum())

print("\nRestaurants missing values:")
print(restaurants.isnull().sum())

print("\nDrivers missing values:")
print(drivers.isnull().sum())

print("\nOrders missing values:")
print(orders.isnull().sum())

print("\nDeliveries missing values:")
print(deliveries.isnull().sum())

print("\nReviews missing values:")
print(reviews.isnull().sum())

print("\nPayments missing values:")
print(payments.isnull().sum())
```

```python
# Check for duplicates
print("Customers missing values:")
print(customers.isnull().sum())

print("\nRestaurants missing values:")
print(restaurants.isnull().sum())

print("\nDrivers missing values:")
print(drivers.isnull().sum())

print("\nOrders missing values:")
print(orders.isnull().sum())

print("\nDeliveries missing values:")
print(deliveries.isnull().sum())

print("\nReviews missing values:")
print(reviews.isnull().sum())

print("\nPayments missing values:")
print(payments.isnull().sum())
```

```python
# Check primary IDs
print("Duplicate customer IDs:",
      customers["customer_id"].duplicated().sum())

print("Duplicate restaurant IDs:",
      restaurants["restaurant_id"].duplicated().sum())

print("Duplicate driver IDs:",
      drivers["driver_id"].duplicated().sum())

print("Duplicate order IDs:",
      orders["order_id"].duplicated().sum())

print("Duplicate delivery IDs:",
      deliveries["delivery_id"].duplicated().sum())

print("Duplicate review IDs:",
      reviews["review_id"].duplicated().sum())

print("Duplicate payment IDs:",
      payments["payment_id"].duplicated().sum())
```

```python
# Check relationship between tables

# Customer -> Orders
invalid_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]
print("Orders with invalid customer IDs:",
      len(invalid_customers))

# Restaurant -> Orders
invalid_restaurants = orders[
    ~orders["restaurant_id"].isin(restaurants["restaurant_id"])
]
print("Orders with invalid restaurant IDs:",
      len(invalid_restaurants))

# Driver -> Deliveries
invalid_drivers = deliveries[
    ~deliveries["driver_id"].isin(drivers["driver_id"])
]
print("Deliveries with invalid driver IDs:",
      len(invalid_drivers))

# Order -> Deliveries
invalid_delivery_orders = deliveries[
    ~deliveries["order_id"].isin(orders["order_id"])
]
print("Deliveries with invalid order IDs:",
      len(invalid_delivery_orders))

# Order -> Reviews
invalid_review_orders = reviews[
    ~reviews["order_id"].isin(orders["order_id"])
]
print("Reviews with invalid order IDs:",
      len(invalid_review_orders))

# Order -> Payments
invalid_payment_orders = payments[
    ~payments["order_id"].isin(orders["order_id"])
]
print("Payments with invalid order IDs:",
      len(invalid_payment_orders))
```

```python
# Check whether deliveries are only for delivered orders
delivery_order_status = deliveries.merge(
    orders[["order_id", "order_status"]],
    on="order_id",
    how="left"
)

print(
    delivery_order_status["order_status"].value_counts()
)
```

```python
# Check numeric ranges

# Customer age
print("Customer Age")
print(customers["age"].describe())

# Restaurant rating
print("Restuarant Rating")
print(restaurants["restaurant_rating"].describe())

# Driver rating
print("Driver Rating")
print(drivers["driver_rating"].describe())

# Order amount
print("Order amount")
print(orders["order_amount"].describe())

# Delivery time
print("Delivery Time")
print(deliveries["delivery_time_minutes"].describe())
```

```python
# Convert the order timestamp
orders["order_timestamp"] = pd.to_datetime(
    orders["order_timestamp"]
)
orders.info()
```

```python
# Final validation summary
print("========== DATA VALIDATION SUMMARY ==========")

print("Customers:", customers.shape)
print("Restaurants:", restaurants.shape)
print("Drivers:", drivers.shape)
print("Orders:", orders.shape)
print("Deliveries:", deliveries.shape)
print("Reviews:", reviews.shape)
print("Payments:", payments.shape)

print("\nTotal missing values:")
print(
    customers.isnull().sum().sum()
    + restaurants.isnull().sum().sum()
    + drivers.isnull().sum().sum()
    + orders.isnull().sum().sum()
    + deliveries.isnull().sum().sum()
    + reviews.isnull().sum().sum()
    + payments.isnull().sum().sum()
)

print("\nValidation completed.")
```
